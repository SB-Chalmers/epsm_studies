# Retain a study before its execution environment is ready
A study protocol and its owner workflow can be saved without engines. This example keeps an intentionally incomplete workflow; it does not claim that it can run. Workflows validates physical and execution choices when they are explicitly requested.

```python
from epsm_studies import Study, StudySpec

spec = StudySpec(
    'Envelope pilot',
    question='What changes annual heating?',
    workflow={'format': 'epsm-workflow/1'},
    protocol={'assumptions': ['The weather selection will be fixed before preparation']},
    reporting={'metrics': ['space_heating_demand_kwh']},
)
study = Study(spec)
assert Study.from_dict(study.to_dict()).id == study.id
study.summary()
```

Saving refuses to overwrite a different record. Inspection returns independent copies, so a notebook edit cannot silently alter the saved protocol. The temporary directory below is only for this example.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as temporary:
    path = study.save(Path(temporary) / 'study.json')
    restored = Study.open(path)
    assert restored.spec.id == spec.id
    restored.spec.reporting['metrics'].append('an edit to an inspection copy')
    assert restored.spec.reporting == spec.reporting
print(restored)
```

To execute a study, install the optional Workflows integration and start its supported template, add subjects/weather, request outputs, inspect preflight, and prepare a frozen plan. After execution, Studies can publish a portable result archive. See the study platform guide for that route and its explicit engine/data prerequisites.
